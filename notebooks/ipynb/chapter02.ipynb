## Aleatorización e inferencia

Un diseño probabilístico permite aprovechar de manera rigurosa los recursos disponibles para estudiar una población biológica. Su característica fundamental es que cada unidad —por ejemplo, un individuo, una parcela, un árbol o un punto de observación— tiene una probabilidad conocida de ser seleccionada. Esto permite utilizar la información obtenida en la muestra para hacer estimaciones sobre toda la población y, además, cuantificar la incertidumbre asociada a esas estimaciones.

No existe un diseño de muestreo que sea el mejor en todas las situaciones. La elección depende de cómo se distribuye la población, de qué tan diferentes son sus unidades, de la información previa disponible, de las condiciones del territorio y de los costos del trabajo de campo [@manly2015ecological; @lohr2022sampling]. Por esta razón, no basta con seleccionar una muestra que visualmente parezca bien distribuida o representativa: es necesario establecer una regla de selección clara, probabilística y reproducible, que permita conocer cómo y con qué probabilidad fue elegida cada unidad.

Consideremos una población finita formada por $N$ unidades, que representamos como $U={1,\ldots,N}$. Para cada unidad $i$, el indicador $I_i$ toma el valor $1$ si la unidad es seleccionada y $0$ si no lo es. Su probabilidad de inclusión se denota por $\pi_i=P(I_i=1)$, y el peso muestral correspondiente es $w_i=1/\pi_i$. Este peso indica cuántas unidades de la población representa aproximadamente cada unidad observada: cuanto menor sea su probabilidad de selección, mayor será su peso.
El estimador de Horvitz–Thompson utiliza estos pesos para estimar el total de una variable $y$ en la población:

$$
\widehat Y_{HT}=\sum_{i\in s}\frac{y_i}{\pi_i}.
$$

En otras palabras, se suma el valor observado de cada unidad seleccionada, ajustado por su probabilidad de inclusión. Por ejemplo, si una unidad tiene una probabilidad de inclusión de $0{,}20$, su peso es $1/0{,}20=5$; por tanto, esa unidad representa cinco unidades de la población.

Para calcular la varianza del estimador también es necesario conocer las probabilidades de inclusión conjunta, $\pi_{ij}$, es decir, la probabilidad de que las unidades $i$ y $j$ sean seleccionadas simultáneamente. Por esta razón, conservar únicamente una columna con los pesos no permite reconstruir por completo el diseño muestral. También deben registrarse los estratos, los conglomerados, las etapas de selección, el orden utilizado y las reglas aplicadas cuando fue necesario reemplazar alguna unidad.
En los ejemplos se estimarán principalmente tres características de la población:

$$
\bar Y=\frac{1}{N}\sum_{i\in U}y_i,\qquad
Y=\sum_{i\in U}y_i,\qquad
P=\frac{1}{N}\sum_{i\in U}z_i.
$$

Aquí, $\bar{Y}$ es la media poblacional, $Y$ es el total poblacional y $P$ es una proporción. Para calcular esta última, $z_i$ toma el valor $1$ cuando la unidad cumple una condición definida previamente y $0$ cuando no la cumple. Finalmente, es importante distinguir la desviación estándar (DE) del error estándar (EE): la DE describe cuánto difieren las unidades de la población entre sí, mientras que el EE indica cuánto podría cambiar una estimación si se repitiera el muestreo. Por tanto, miden tipos de variabilidad diferentes y no deben utilizarse como sinónimos.


## Diseños fundamentales

### Muestreo aleatorio simple

En un muestreo aleatorio simple sin reemplazo (MAS), se seleccionan \(n\) unidades de una población de tamaño \(N\), de manera que todas las muestras posibles de ese tamaño tienen la misma probabilidad de ser elegidas. La expresión «sin reemplazo» significa que una unidad seleccionada no puede volver a aparecer en la misma muestra. En este diseño, la media de la población se estima mediante la media de las unidades seleccionadas:

$$
\widehat{\bar Y}=\bar y_s=\frac{1}{n}\sum_{i\in s}y_i.
$$

El numerador suma solo los $n$ valores observados, por lo que se divide entre $n$,
el tamaño de la muestra, y no entre $N$. Esta media estima la media poblacional
desconocida $\bar Y=(1/N)\sum_U y_i$: ambas suman sus términos y los dividen entre
su propio número (la muestra entre $n$; la población, entre $N$, que no se conoce
porque solo se observó la muestra). Si se dividiera por $N$ se mezclaría una suma
parcial con el total de unidades de la población.

La variabilidad entre las unidades de la muestra se resume con la varianza
muestral

$$
s_y^2=\frac{1}{n-1}\sum_{i\in s}\left(y_i-\bar y_s\right)^2,
$$

que se divide entre $n-1$ porque estima la varianza poblacional $S^2$ ajustando
un grado de libertad. Para cuantificar cuánto varía *la media estimada* entre
muestras repetidas, esa variabilidad entre unidades se divide además entre $n$,
pues promediar $n$ valores reduce la variación en un factor $1/n$ (de ahí el
error estándar $SE=\sqrt{s_y^2/n}=s_y/\sqrt n$). La varianza de la media es

$$
\widehat V(\bar y_s)=\left(1-\frac nN\right)\frac{s_y^2}{n},
$$

donde el factor $1-n/N$, la corrección por población finita, reduce la varianza
cuando la muestra cubre una fracción importante de la población. Si $n/N$ es
pequeña casi no afecta; si $n=N$ se observa todo y la varianza es cero. En
resumen: el $n-1$ vive dentro de $s_y^2$ (variabilidad entre unidades) y el $n$
divide en la varianza de la media (efecto de promediar).
A partir de la media muestral también pueden estimarse otras características. El total de la población se estima como

$$
\widehat Y=N\bar y_s.
$$

Así, en el MAS cada cantidad tiene su papel: la *media* se estima dividiendo la
suma entre $n$; la *varianza entre unidades* $s_y^2$ se estima dividiendo entre
$n-1$; la *varianza de la media* divide esa $s_y^2$ entre $n$ y la ajusta con
$1-n/N$; y el *total* multiplica la media por $N$. $N$ no aparece en la media,
sino al extenderla a toda la población y en la corrección por finitud.

Para estimar una proporción, se define una variable indicadora que toma el valor $1$ si cada unidad cumple la condición de interés y $0$ si no la cumple. La media de ese indicador en la muestra corresponde a la proporción estimada, y su varianza se calcula siguiendo el mismo principio.

### Muestreo con probabilidad proporcional al tamaño (PPS)

El MAS da a todas las unidades la misma probabilidad de entrar en la muestra.
A veces eso no conviene: si algunas unidades son mucho más grandes o más
importantes (por ejemplo, los claros de mayor tamaño en un bosque), puede
interesar que esas unidades tengan más oportunidades de ser seleccionadas. En el
muestreo con probabilidad proporcional al tamaño, o PPS, la probabilidad de cada
unidad es proporcional a una medida de tamaño conocida $x_i$ para todo el marco:

$$
\pi_i=n\frac{x_i}{X},\qquad X=\sum_{i\in U}x_i,
$$

donde $n$ es el tamaño de muestra y $X$ el total de la medida de tamaño. Una
unidad con el doble de $x_i$ tiene el doble de probabilidad de ser elegida.

Con el estimador de Horvitz–Thompson $\widehat Y_{HT}=\sum_{s}y_i/\pi_i$, el
total estimado es insesgado aunque las probabilidades sean desiguales. Si la
variable $y$ está relacionada con el tamaño $x$, PPS suele dar mayor precisión
que el MAS y hace que cada unidad seleccionada tenga una contribución similar al
resultado.

Hay dos formas habituales de llevarlo a la práctica.

*Con reemplazo*: se hacen $n$ sorteos independientes y, en cada sorteo, la
unidad $i$ se elige con probabilidad $q_i=x_i/X$. El procedimiento es simple,
aunque una misma unidad puede entrar varias veces. Si $z_k=y_{i_k}/q_i$ es el
valor expandido del sorteo $k$, el total se estima con la media de esos $z_k$ y su
varianza se estima sin probabilidades conjuntas, solo con la dispersión de los
valores sorteados:

$$
\widehat Y_{pps}=\frac1n\sum_{k=1}^{n}z_k,\qquad
\widehat V(\widehat Y_{pps})=\frac{1}{n(n-1)}\sum_{k=1}^{n}\left(z_k-\widehat Y_{pps}\right)^2.
$$

*PPS sistemático o por tamaños acumulados*: se ordenan las unidades, se calculan
los totales acumulados de $x$ y se colocan, cada intervalo fijo, un punto de
selección sobre un arranque aleatorio. Aproxima un PPS sin reemplazo, es muy
práctico en campo y comparte con el muestreo sistemático el mismo riesgo de
alineación: si el orden de $x$ tiene periodicidad, la selección puede quedar
sesgada.

PPS no es un estimador en sí: usa el tamaño $x$ en la *selección*, igual que el
estimador de razón usa $X$ en la *estimación* (sección siguiente). Exige una
medida de tamaño precisa para todo el marco; si $x$ predice mal a $y$, la ventaja
desaparece e incluso puede perder precisión frente al MAS. El PPS estricto sin
reemplazo obliga a probabilidades conjuntas $\pi_{ij}$ difíciles de obtener, por
lo que en la práctica se usan aproximaciones como el PPS sistemático.

Cuando $X$ o el tamaño de la población no se conocen bien, conviene el estimador
de Hájek, un estimador de razón que estabiliza el total expandido dividiendo por
el tamaño estimado con los mismos pesos:

$$
\widehat{\bar y}_H=\frac{\sum_{s}w_i y_i}{\sum_{s}w_i},\qquad
\widehat Y_H=N\,\widehat{\bar y}_H=N\frac{\sum_{s}w_i y_i}{\sum_{s}w_i},
$$

donde $w_i=1/\pi_i$. Con pesos muy variables, el cociente multiplica menos que
$\widehat Y_{HT}$, reduce la inflación que provocaría un solo peso grande y
mejora la estabilidad; a cambio, su esperanza no es exactamente insesgada, con
un sesgo pequeño como el de todo estimador de razón.

### Muestreo estratificado

En vez de tomar toda la muestra de golpe, la población se divide antes de seleccionar en grupos llamados *estratos* que no se solapan. La idea es que cada estrato sea lo más homogéneo por dentro y lo más distinto de los demás: por ejemplo bosque, matorral y pastizal. Dentro de cada estrato se toma una muestra aleatoria propia de tamaño $n_h$.

Cada estrato $h$ aporta $N_h$ unidades de las $N$ totales; su *peso* es $W_h=N_h/N$, la fracción de la población que vive en él. Estimamos la media de cada estrato con la media de su muestra y las combinamos:

$$
\widehat{\bar{Y}}_{st}=\sum_hW_h\bar y_h,\qquad
\widehat V(\widehat{\bar{Y}}_{st})=
\sum_hW_h^2\left(1-\frac{n_h}{N_h}\right)\frac{s_h^2}{n_h}.
$$

El estimador es un promedio ponderado: un estrato con más unidades aporta más al
resultado. La varianza estimada suma la de cada estrato, cada una multiplicada
por el cuadrado de su peso; el factor $1-n_h/N_h$ es la corrección por población
finita aplicada dentro de cada estrato.

**Cómo repartir la muestra $n$ entre estratos.**

*Proporcional*: se reparte según el tamaño. Si el bosque contiene la mitad de las
parcelas, recibe la mitad de la muestra, $n_h\approx nN_h/N$. Simple y justa: cada
unidad tiene la misma probabilidad de salir. Conviene cuando todos los estratos
son tan variables o "parecidos" entre sí.

*Neyman*: se reparte por tamaño $y$ heterogeneidad, $n_h\propto N_hS_h$, donde
$S_h$ es la desviación de cada estrato. Así damos más muestra a los estratos
grandes y a los más variables, que son donde hay más incertidumbre que reducir.
Esta regla minimiza la varianza cuando el costo por unidad es igual en todos los
estratos, pero exige conocer $S_h$ de antemano, de un estudio previo o un piloto.

Estratificar garantiza observaciones en dominios pequeños y, si los estratos son
internamente homogéneos, puede dar un error menor que el MAS. La condición
indispensable es poder asignar cada unidad del marco a su estrato *antes* de
seleccionar.


### Muestreo sistemático

Si la población está ordenada en una lista, podemos muestrear con un *arranque
aleatorio* y *saltos regulares*: elegimos un punto inicial al azar y, a partir de
él, tomamos una unidad cada $k$ posiciones (por ejemplo, una parcela cada diez).
El esfuerzo queda repartido a lo largo de todo el marco, no concentrado en una
zona.

Su punto débil es el *orden* de la lista. Si el orden es un gradiente suave
(diámetros que crecen de principio a fin), la cobertura regular captura
representantes de todo el gradiente y puede ser más precisa que el MAS. Pero si
la serie tiene periodicidad (eventos que se repiten cada $k$ posiciones), la
selección puede alinearse con ella y sobrestimar o subestimar de modo sistemático.
Por eso la precisión no depende solo del azar: depende de cómo estén ordenadas
las unidades.

Una desventaja adicional es que una sola secuencia no ofrece, en general, un
estimador interno de su propia varianza: no tenemos varias muestras con las que
medir cuánto variaría el resultado. Para cuantificar incertidumbre cabe usar
*arranques independientes* (repetir el diseño varias veces), una *aproximación
declarada* (como suponer el orden aleatorio y usar la fórmula del MAS) o un
*análisis basado en la estructura espacial* de los datos.

### Conglomerados y etapas

En vez de elegir unidades sueltas, puede convenir seleccionar *grupos* enteros
llamados conglomerados, por ejemplo las parcelas de un mismo bosque. Hay dos
variantes. En *una etapa* se seleccionan los grupos y luego se observan todas sus
unidades. En *dos etapas* se seleccionan primero algunos grupos y, dentro de cada
grupo elegido, solo unas cuantas unidades. La ventaja es logística: viajar y
medir en un mismo sitio es más barato y práctico que saltar de parcela en parcela.

La desventaja es estadística. Las unidades de un mismo conglomerado suelen
parecerse entre sí: dos parcelas vecinas ofrecen los mismos árboles, humedad y
suelo. Cuando las unidades repetidas son muy semejantes, cada observación nueva
aporta poca información que no tuviéramos ya; en la práctica, medir bastantes
unidades dentro de un grupo equivale a *casi una sola* medida. Cuanto más parecidas
sean, menor será la información efectiva de la muestra. La correlación intraclase,
$\rho$, mide precisamente ese parecido interno: vale $0$ si las unidades no se
parecen y se acerca a $1$ si son casi idénticas.

El *efecto de diseño* resume cuánto pierde (o gana) el diseño frente al MAS con
el mismo número de observaciones:

$$
DEFF\approx1+(L-1)\rho,
$$

donde $L$ es el tamaño del conglomerado y $\rho$ la correlación intraclase. Si
$\rho>0$, el $DEFF>1$: se necesitan más observaciones que en un MAS para lograr
la misma precisión. La clave es que quien fija la precisión es el número de
grupos seleccionados (las unidades primarias), no el de observaciones internas:
replicar dentro de un grupo ya muestreado casi no reduce la incertidumbre del
total.

### Información auxiliar

A veces conocemos, para todo el marco, una variable $x_i$ distinta de la que
estudiamos, por ejemplo la medida registrada en el censo, más barata o más fácil
de obtener que $y_i$. Cuando esa variable *se relaciona* con $y_i$, podemos usarla
para estimar mejor el total sin necesidad de medir $y_i$ en todas las unidades.

La idea es comparar proporciones. Con una muestra aleatoria estimamos cuánto vale
$y_i$ por cada unidad de $x_i$ como el cociente

$$r=\frac{\sum_{i\in s}y_i}{\sum_{i\in s}x_i},$$

la "razón" media en la muestra. Si sabemos el total de $x$ en toda la población,
$X=\sum_{i\in U}x_i$, un total razonable de $y$ es

$$\widehat Y_R=r\,X=
\frac{\sum_{i\in s}y_i}{\sum_{i\in s}x_i}X.$$

Piénselo como una regla de tres: si la muestra dice que por cada unidad de
medida hay $r$ unidades de $y$, y el marco tiene $X$ unidades en total, el total
de $y$ será alrededor de $rX$.

Este método gana precisión cuando la relación entre $y$ y $x$ es *estable y muy
cercana al origen* (un punto con $x=0$ corresponde a $y\approx0$): entonces el
cociente $r$ apenas cambia de una muestra a otra. Si la relación tiene un
intercepto marcado o es curva, el cociente funciona peor. Además, su sesgo no es
exactamente cero: al no ser una combinación lineal simple del diseño, debe
compararse con el estimador que solo depende de la muestra del MAS, y decidir
cuál conviene revisando residuos y precisión.

## Población finita de árboles

`datasets::trees` contiene diámetro, altura y volumen de 31 cerezos negros
talados. Las unidades documentadas son pulgadas para `Girth`, pies para `Height`
y pies cúbicos para `Volume`; pese al nombre histórico, `Girth` representa
diámetro. No se incluyen ubicación, fecha ni mecanismo original de selección.
Las 31 filas se tratarán como una población finita real para estudiar
aleatorización, sin generalizar a un bosque actual.

El estimando principal es el volumen total. También estimaremos el volumen medio,
su DE finita y la proporción de árboles con al menos 30 pies cúbicos. El umbral se
fija antes de tomar las muestras. Para estimación auxiliar se usará
$x=\text{diámetro}^2\times\text{altura}$, conocido para todo el marco del
ejercicio.

### Carga, procedencia y auditoría

```r
data(trees, package = "datasets")
U <- transform(trees, id = seq_len(nrow(trees)), grande = as.integer(Volume >= 30), x_aux = Girth^2 * Height)
N <- nrow(U)
n <- 10

# La DE poblacional describe la variabilidad entre unidades del marco. Se expresa
# con el divisor finito N-1: equivale a sd()*sqrt((N-1)/N) y no debe confundirse
# con el EE del estimador.
verdad_arboles <- c(media = mean(U$Volume), total = sum(U$Volume), proporcion = mean(U$grande), DE = sd(U$Volume) * sqrt((N - 1) / N))

data.frame(
  N = N,
  ids_unicos = length(unique(U$id)),
  faltantes = sum(is.na(U)),
  duplicados = sum(duplicated(trees)),
  valores_no_positivos = sum(U[c("Girth", "Height", "Volume")] <= 0),
  correlacion_auxiliar = cor(U$x_aux, U$Volume)
)
round(verdad_arboles, 3)
```

No aparecen faltantes, duplicados exactos ni valores no positivos. La auditoría
no puede detectar sesgo de cobertura porque el marco y la selección histórica no
están documentados. La correlación alta es una propiedad de estas 31 filas, no
una garantía para otra población.

## Ejemplo 1: muestreo aleatorio simple

Seleccionamos diez árboles sin reemplazo. Cada árbol tiene
$\pi_i=n/N=10/31$ y peso $w_i=3.1$.

```r
set.seed(2201)
ids_mas <- sample(U$id, n)
s_mas <- U[match(ids_mas, U$id), ]

est_mas <- c(
  media = mean(s_mas$Volume),
  total = N * mean(s_mas$Volume),
  proporcion = mean(s_mas$grande),
  DE = sd(s_mas$Volume)
)
se_mas <- c(
  media = sqrt((1 - n / N) * var(s_mas$Volume) / n),
  total = N * sqrt((1 - n / N) * var(s_mas$Volume) / n),
  proporcion = sqrt((1 - n / N) * var(s_mas$grande) / n)
)
t_mas <- qt(0.975, n - 1) # t de student (valor crítico) con n-1=9 grados de libertad
ic_mas <- cbind(
  LI = est_mas[c("media", "total", "proporcion")] - t_mas * se_mas,
  LS = est_mas[c("media", "total", "proporcion")] + t_mas * se_mas
)
ic_mas["proporcion", ] <- pmin(1, pmax(0, ic_mas["proporcion", ])) 

round(cbind(
  estimacion = est_mas[c("media", "total", "proporcion")],
  EE = se_mas,
  ic_mas
), 3)
round(est_mas["DE"], 3)
```

La DE describe diferencias de volumen entre árboles; el EE de la media disminuye
además con $n$ y con la corrección finita. El intervalo de la proporción es una
aproximación basada en *t* y se restringe al rango posible; con conteos pequeños
conviene no confiar únicamente en esa aproximación.

La selección aleatoria puede dejar vacías regiones visuales del diagrama. Eso no

realización.

```r
#| label: fig-mas-distribucion
#| fig-cap: "Distribución del volumen en la población y en la muestra MAS."
breaks_v <- pretty(range(U$Volume), n = 7)
hist(U$Volume, breaks = breaks_v, freq = FALSE, col = "grey90",
     border = "white", xlab = "Volumen (pies cúbicos)", main = "")
hist(s_mas$Volume, breaks = breaks_v, freq = FALSE, add = TRUE,
     col = adjustcolor("#bc4749", 0.55), border = "white")
legend("topright", c("Población", "Muestra"), fill = c("grey90", "#bc4749"),
       bty = "n")
```

```r
#| label: fig-mas-intervalos
#| fig-cap: "Estimaciones MAS e intervalos del 95% frente a los valores de la población finita."
op <- par(mfrow = c(1, 3), mar = c(4, 4, 2, 1))
for (j in seq_along(se_mas)) {
  nombre <- names(se_mas)[j]
  plot(1, est_mas[nombre], ylim = range(c(ic_mas[nombre, ],
       verdad_arboles[nombre])), xaxt = "n", xlab = "", ylab = nombre,
       pch = 19, col = "#386641")
  arrows(1, ic_mas[nombre, 1], 1, ic_mas[nombre, 2], angle = 90,
         code = 3, length = 0.08)
  abline(h = verdad_arboles[nombre], lty = 2, col = "#bc4749")
}
par(op)
```

### Evaluación por repetición

Repetir el código no añade información a la muestra observada. Aquí es posible
repetir el diseño porque se conocen los 31 valores, y así comprobar sus
propiedades de largo plazo.

```r
set.seed(2202)
B <- 3000
rep_mas <- t(replicate(B, {
  z <- U[sample.int(N, n), ]
  est <- c(media = mean(z$Volume), total = N * mean(z$Volume),
           proporcion = mean(z$grande))
  se <- c(media = sqrt((1 - n / N) * var(z$Volume) / n),
          total = N * sqrt((1 - n / N) * var(z$Volume) / n),
          proporcion = sqrt((1 - n / N) * var(z$grande) / n))
  c(est,
    se_media = unname(se["media"]),
    se_total = unname(se["total"]),
    se_proporcion = unname(se["proporcion"]),
    cubre_media = unname(abs(est["media"] - verdad_arboles["media"]) <=
      t_mas * se["media"]),
    cubre_total = unname(abs(est["total"] - verdad_arboles["total"]) <=
      t_mas * se["total"]),
    cubre_proporcion = unname(abs(est["proporcion"] - verdad_arboles["proporcion"]) <=
      t_mas * se["proporcion"]))
}))

metricas_repeticion <- function(x, verdad) c(
  sesgo = mean(x - verdad),
  DE_empirica = sd(x),
  RMSE = sqrt(mean((x - verdad)^2))
)

# Cada repetición guarda también su error estándar y si su intervalo cubre el
# valor verdadero; así se suma el SE medio y la cobertura al diagnóstico.
metricas_repeticion_ic <- function(est, se, cubre, verdad) c(
  sesgo = mean(est - verdad),
  DE_empirica = sd(est),
  SE_medio = mean(se),
  RMSE = sqrt(mean((est - verdad)^2)),
  cobertura = mean(cubre)
)
rbind(
  media = metricas_repeticion_ic(rep_mas[, "media"], rep_mas[, "se_media"],
      rep_mas[, "cubre_media"], verdad_arboles["media"]),
  total = metricas_repeticion_ic(rep_mas[, "total"], rep_mas[, "se_total"],
      rep_mas[, "cubre_total"], verdad_arboles["total"]),
  proporcion = metricas_repeticion_ic(rep_mas[, "proporcion"],
      rep_mas[, "se_proporcion"], rep_mas[, "cubre_proporcion"],
      verdad_arboles["proporcion"])
)
```

Al repetir el diseño 3000 veces, los totales estimados se agrupan alrededor del valor poblacional verdadero (935.3 pies cúbicos), con un sesgo de 1.16 frente a una DE de 129.35: el estimador no tiende a errar en una dirección sistemática, y el ancho de la campana es la incertidumbre real del estimador. Ese diagnóstico se cierra comparando las dos fuentes de esa incertidumbre. La DE empírica es la dispersión observada de los 3000 totales; el error estándar medio es el promedio del error estándar que predice la fórmula $SE=N\sqrt{(1-n/N)s^2/n}$ en una muestra. Que coincidan (129.35 frente a 129.40) valida la fórmula: la variación estimada a partir de una sola muestra describe bien la variación real del total. La cobertura extiende ese resultado al intervalo: en ~94 de cada 100 repeticiones el intervalo del 95% capturó el total verdadero (y ~0.96 en el caso de la proporción, cuya aproximación normal con diez observaciones sigue siendo razonable). ## Ejemplo 2: muestreo estratificado

Los estratos se definen con diámetro, disponible para todo el marco antes de
observar la muestra. Los puntos de corte corresponden a terciles del marco y se
conservan como parte del diseño. La muestra de $n=10$ se reparte entre estratos
por asignación proporcional al tamaño ($n_h\approx nN_h/N$), redondeando por
restos mayores; se imprime también la asignación de Neyman, que en campo exigiría
conocer $S_h$ de un piloto.

```r
cortes <- c(8.3, 11.3, 14.0, 20.6)
U$estrato <- cut(U$Girth, cortes, include.lowest = TRUE,
                 labels = c("pequeno", "medio", "grande"))
Nh <- table(U$estrato)
n_total <- 10

asignar_enteros <- function(pesos, n_total, minimo = 2) {
  a <- rep(minimo, length(pesos)); names(a) <- names(pesos)
  faltan <- n_total - sum(a)
  if (faltan > 0) {
    cuota <- faltan * pesos / sum(pesos)
    a <- a + floor(cuota)
    resto <- n_total - sum(a)
    if (resto > 0) {
      fraccion <- cuota - floor(cuota)
      idx <- order(fraccion, decreasing = TRUE)[seq_len(resto)]
      a[idx] <- a[idx] + 1
    }
  }
  a
}

# Asignación proporcional: n_h ≈ n·N_h/N
nh <- asignar_enteros(as.numeric(Nh), n_total)
names(nh) <- names(Nh)
# Asignación de Neyman: n_h ∝ N_h·S_h (referencia; exigiría S_h previa)
S_h <- tapply(U$Volume, U$estrato, sd)
nh_neyman <- asignar_enteros(as.numeric(Nh) * S_h, n_total)
names(nh_neyman) <- names(Nh)
round(rbind(proporcional = nh, Neyman_referencia = nh_neyman, poblacion = Nh), 0)

set.seed(2203)
ids_st <- unlist(lapply(names(nh), function(h) {
  sample(U$id[U$estrato == h], nh[h])
}))
s_st <- U[match(ids_st, U$id), ]
table(poblacion = U$estrato)
table(muestra = s_st$estrato)
```

La asignación proporcional reparte según el tamaño del estrato: como el estrato
pequeño contiene 11 de los 31 árboles, recibe cuatro de las diez observaciones.
La de Neyman, en cambio, pesa cada estrato por su desviación $S_h$; aquí concentra
cuatro observaciones en el estrato grande, que es el más variable. Ningún tamaño
se fija a mano: ambos salen de la misma regla de restos mayores.

El estimador de estrato se aplica tanto al volumen como a cualquier otra variable, incluido el indicador. Para resumir la variabilidad entre unidades (la DE de la población) no basta con mirar una muestra: se estiman los totales de $y$ y de $y^2$ y, a partir de ellos, la varianza poblacional $S^2$. Como el cuadrado de un estimador no estima sin sesgo el cuadrado del parámetro —ya que $E(\widehat Y^{\,2})=Y^2+\operatorname{Var}(\widehat Y)$—, al cuadrado del total estimado se le resta su varianza estimada $\widehat V(\widehat Y)$ antes de combinarlo con el total de $y^2$; de lo contrario se sobreestimaría $Y^2$ y, con ello, la DE estimada.

```r
estimar_st <- function(datos, variable, Nh, nh) {
  por_h <- lapply(names(nh), function(h) {
    z <- datos[as.character(datos$estrato) == h, variable]
    N_h <- unname(Nh[h])
    n_h <- unname(nh[h])
    c(total = N_h * mean(z),
      var_total = N_h^2 * (1 - n_h / N_h) * var(z) / n_h)
  })
  por_h <- do.call(rbind, por_h)
  c(total = sum(por_h[, "total"]), var_total = sum(por_h[, "var_total"]))
}

st_y <- estimar_st(s_st, "Volume", Nh, nh)
st_y2 <- estimar_st(transform(s_st, Volume2 = Volume^2), "Volume2", Nh, nh)
st_p <- estimar_st(s_st, "grande", Nh, nh)
S2_st <- (st_y2["total"] -
            (st_y["total"]^2 - st_y["var_total"]) / N) / (N - 1)

est_st <- c(media = unname(st_y["total"] / N),
            total = unname(st_y["total"]),
            proporcion = unname(st_p["total"] / N),
            DE = sqrt(max(0, S2_st)))
se_st <- c(media = unname(sqrt(st_y["var_total"]) / N),
           total = unname(sqrt(st_y["var_total"])),
           proporcion = unname(sqrt(st_p["var_total"]) / N))
t_st <- qt(0.975, n - length(nh))
ic_st <- cbind(LI = est_st[names(se_st)] - t_st * se_st,
               LS = est_st[names(se_st)] + t_st * se_st)
ic_st["proporcion", ] <- pmin(1, pmax(0, ic_st["proporcion", ]))
round(cbind(estimacion = est_st[names(se_st)], EE = se_st, ic_st), 3)
round(est_st["DE"], 3)
```

Los resultados del muestreo estratificado son coherentes con la población finita conocida. El total estimado, 906.9 pies cúbicos, queda cerca del valor poblacional verdadero (935.3), y su intervalo del 95% $[693.5, 1120.3]$ lo contiene de sobra; lo mismo ocurre con la media (29.25 estimada frente a ~30.17 verdadera, IC $[22.4, 36.1]$). El error estándar del total (90.3) refleja la reducción de incertidumbre que aporta la estratificación frente a un MAS del mismo tamaño al explotar la homogeneidad interna de cada estrato. La proporción de árboles con volumen alto se estima en 0.43 con un intervalo ancho $[0.22, 0.64]$, consecuencia de que, con solo diez observaciones, la precisión para ese indicador binario es limitada. Por último, la desviación estándar entre unidades se estima en 16.60 pies cúbicos, un valor intermedio que describe la variabilidad individual de los árboles y difiere, por definición, del error estándar del estimador.

El total se obtiene sumando los totales de estrato. Promediar sin pesos las tres
medias sería incorrecto porque los tamaños $N_h$ difieren.

### Asignación y repetición del diseño

$nN_h/N$. La asignación de Neyman usa las DE censales de volumen únicamente como
referencia retrospectiva; en campo deberían proceder de un piloto independiente.
Se reutiliza la función `asignar_enteros` ya definida.

```r
nh_prop <- asignar_enteros(as.numeric(Nh), n); names(nh_prop) <- names(Nh)
S_h_censal <- tapply(U$Volume, U$estrato, sd)
nh_neyman <- asignar_enteros(as.numeric(Nh) * S_h_censal, n)
names(nh_neyman) <- names(Nh)
rbind(proporcional = nh_prop, Neyman_retrospectiva = nh_neyman)

una_st <- function(nh_local) {
  ids <- unlist(lapply(names(nh_local), function(h) {
    sample(U$id[U$estrato == h], nh_local[h])
  }))
  z <- U[match(ids, U$id), ]
  estimar_st(z, "Volume", Nh, nh_local)["total"]
}

set.seed(2204)
rep_st <- cbind(
  proporcional = replicate(B, una_st(nh_prop)),
  Neyman = replicate(B, una_st(nh_neyman))
)
metricas_st <- rbind(
  MAS = metricas_repeticion(rep_mas[, "total"], verdad_arboles["total"]),
  proporcional = metricas_repeticion(rep_st[, "proporcional"],
                                     verdad_arboles["total"]),
  Neyman_retrospectiva = metricas_repeticion(rep_st[, "Neyman"],
                                             verdad_arboles["total"])
)
round(metricas_st, 3)
```

## Ejemplo 3: muestreo sistemático

Ordenamos el marco por diámetro y usamos diez posiciones circulares casi
equiespaciadas. Los saltos alternan entre tres y cuatro posiciones porque 31 no
es divisible por 10. El arranque tiene 31 valores equiprobables y cada árbol
aparece en diez de las secuencias.

```r
Uo <- U[order(U$Girth, U$id), ]
seleccion_sistematica <- function(arranque, n = 10, N = nrow(Uo)) {
  desplazamientos <- floor((0:(n - 1)) * N / n)
  ((arranque - 1 + desplazamientos) %% N) + 1
}

set.seed(2205)
arranque <- sample.int(N, 1)
pos_sys <- seleccion_sistematica(arranque, n, N)
s_sys <- Uo[pos_sys, ]
est_sys <- c(
  media = mean(s_sys$Volume),
  total = N * mean(s_sys$Volume),
  proporcion = mean(s_sys$grande),
  DE = sd(s_sys$Volume)
)
c(arranque = arranque, ids = paste(s_sys$id, collapse = ", "))
round(est_sys, 3)
```

### Distribución exacta de arranques

Como la población didáctica está completa, podemos enumerar los 31 arranques.
Esto proporciona el EE de diseño exacto para este marco, pero no un estimador
operativo disponible a partir de una sola secuencia en campo.

```r
todos_sys <- t(vapply(seq_len(N), function(a) {
  z <- Uo[seleccion_sistematica(a, n, N), ]
  c(media = mean(z$Volume), total = N * mean(z$Volume),
    proporcion = mean(z$grande))
}, numeric(3)))
de_diseno_exacta <- function(x) sqrt(mean((x - mean(x))^2))
se_sys_diseno <- apply(todos_sys, 2, de_diseno_exacta)
ic_sys_referencia <- cbind(
  LI = est_sys[names(se_sys_diseno)] - 1.96 * se_sys_diseno,
  LS = est_sys[names(se_sys_diseno)] + 1.96 * se_sys_diseno
)
ic_sys_referencia["proporcion", ] <- pmin(1,
  pmax(0, ic_sys_referencia["proporcion", ]))
round(cbind(estimacion = est_sys[names(se_sys_diseno)],
            EE_diseno = se_sys_diseno, ic_sys_referencia), 3)

# Sin enumerar arranques, la única manera operativa de declarar la incertidumbre
# desde una sola secuencia es tratarla como un MAS: N*sqrt((1-n/N)*s^2/n).
se_sys_declarado <- N * sqrt((1 - n / N) * var(s_sys$Volume) / n)
c(EE_declarado_como_MAS = se_sys_declarado)
```

Enumerar los 31 arranques permite evaluar el diseño sin aproximaciones: el
promedio de los totales de todos los arranques es exactamente el total
poblacional (935.3), lo que confirma que la sistemática es insesgada, y su
desviación entre arranques es de 67.06 pies cúbicos. El total de la secuencia
sorteada (1009.05, arranque 31) cae dentro del rango que recorren los 31
arranques (797.6 a 1030.4): es uno más de los resultados equiprobables. Este EE
de diseño (67.06) es notablemente menor que la desviación empírica del total
MAS (129.35); al ordenar por un gradiente, la sistemática reduce a la mitad la
variabilidad del mismo tamaño muestral. El intervalo asociado es una referencia
de evaluación que usa todos los arranques y una aproximación normal; no debe
presentarse como calculable desde la única secuencia realizada en campo.

Esa salvedad se ve con la declaración operativa. Si de una sola secuencia se
aplicara la fórmula del MAS —la única disponible sin repetir diseños—, el error
estándar declarado (162.55) casi duplicaría el de diseño (67.06): tratar la
sistemática como un MAS paga la variabilidad completa del gradiente y sobreestima
la incertidumbre. La lección es que la sistemática gana precisión exactamente en
el rasgo que la fórmula del MAS ignora (la cobertura regular del orden), así que
su varianza no es directamente estimable desde una sola secuencia; en campo se
aproxima con dos o más sistemas independientes, como se discute en los capítulos
de diseño.

## Ejemplo 4: conglomerados y dos etapas

`vegan::BCI` contiene conteos de árboles con DAP de al menos 10 cm en 50
hectáreas contiguas de Barro Colorado, Panamá [@condit2002beta;
@harms2001habitat; @pyke2001floristic; @oksanen2024vegan]. Aquí se reutiliza el
censo únicamente para representar una jerarquía de selección; el estudio de
soporte, comunidad y dependencia espacial corresponde al capítulo anterior.

La población finita son las 50 hectáreas. Las diez columnas espaciales serán
unidades primarias o conglomerados, cada una con cinco hectáreas. La respuesta es
el número de tallos por hectárea; también se estima la proporción de hectáreas con
riqueza de al menos 95 especies.

```r
data(BCI, BCI.env, package = "vegan")
Y_bci <- as.matrix(BCI)
marco_bci <- BCI.env
marco_bci$id <- seq_len(nrow(marco_bci))
marco_bci$x <- as.integer((marco_bci$UTM.EW - min(marco_bci$UTM.EW)) / 100 + 1)
marco_bci$y <- as.integer((marco_bci$UTM.NS - min(marco_bci$UTM.NS)) / 100 + 1)
marco_bci$conglomerado <- marco_bci$x
marco_bci$tallos <- rowSums(Y_bci)
marco_bci$riqueza <- vegan::specnumber(Y_bci)
marco_bci$riqueza_alta <- as.integer(marco_bci$riqueza >= 95)

M <- length(unique(marco_bci$conglomerado))
L <- as.integer(table(marco_bci$conglomerado)[1])
N_bci <- nrow(marco_bci)
m <- 4
l <- 3

data.frame(
  hectareas = N_bci,
  conglomerados = M,
  hectareas_por_conglomerado = L,
  faltantes = sum(is.na(cbind(Y_bci, marco_bci[c("x", "y")]))),
  coordenadas_duplicadas = sum(duplicated(marco_bci[c("x", "y")]))
)
```

### Una etapa: seleccionar franjas completas

Con totales de conglomerado $t_h$, un MAS de $m$ entre $M$ conglomerados estima

$$
\widehat Y_c=\frac{M}{m}\sum_{h\in s}t_h,
\qquad
\widehat V(\widehat Y_c)=
M^2\left(1-\frac mM\right)\frac{s_t^2}{m}.
$$

```r
set.seed(2206)
clusters_s <- sort(sample(seq_len(M), m))
s_cluster <- marco_bci[marco_bci$conglomerado %in% clusters_s, ]

estimar_cluster <- function(datos, variable, M, m, N_total) {
  totales <- tapply(datos[[variable]], datos$conglomerado, sum)
  est_total <- M / m * sum(totales)
  var_total <- M^2 * (1 - m / M) * var(totales) / m
  c(total = est_total, var_total = var_total, media = est_total / N_total,
    EE_media = sqrt(var_total) / N_total)
}

cl_y <- estimar_cluster(s_cluster, "tallos", M, m, N_bci)
cl_y2 <- estimar_cluster(transform(s_cluster, tallos2 = tallos^2),
                         "tallos2", M, m, N_bci)
cl_p <- estimar_cluster(s_cluster, "riqueza_alta", M, m, N_bci)
S2_cl <- (cl_y2["total"] - (cl_y["total"]^2 - cl_y["var_total"]) / N_bci) /
  (N_bci - 1)
t_cl <- qt(0.975, m - 1)
res_cl <- rbind(
  media_tallos = c(estimacion = unname(cl_y["media"]),
                   EE = unname(cl_y["EE_media"])),
  total_tallos = c(estimacion = unname(cl_y["total"]),
                   EE = unname(sqrt(cl_y["var_total"]))),
  proporcion_riqueza_alta = c(estimacion = unname(cl_p["media"]),
                              EE = unname(cl_p["EE_media"]))
)
res_cl <- cbind(res_cl, LI = res_cl[, "estimacion"] - t_cl * res_cl[, "EE"],
                LS = res_cl[, "estimacion"] + t_cl * res_cl[, "EE"])
res_cl["proporcion_riqueza_alta", c("LI", "LS")] <- pmin(1, pmax(0,
  res_cl["proporcion_riqueza_alta", c("LI", "LS")]))
round(res_cl, 3)
c(DE_tallos = sqrt(max(0, S2_cl)), conglomerados = paste(clusters_s, collapse = ","))
```

Con $m=4$ las medias puntuales (425.3, EE 5.44) quedan cerca de la verdad
censal (429.1), pero los grados de libertad son $m-1=3$: observar 20 hectáreas
no crea 20 réplicas independientes cuando solo se seleccionaron cuatro unidades
primarias. El EE de diseño (5.44) corresponde a una realización; con pocos
conglomerados la variabilidad de largo plazo (DE empírica ≈ 10.3) puede
excederlo.

### Dos etapas: seleccionar franjas y hectáreas

En cada conglomerado seleccionado se toman tres de cinco hectáreas. Con
$\widehat t_h=L\bar y_h$ y
$\widehat V_h=L^2(1-l/L)s_h^2/l$,

$$
\widehat Y_{2e}=\frac{M}{m}\sum_{h\in s}\widehat t_h,
$$

$$
\widehat V(\widehat Y_{2e})=
M^2\left(1-\frac mM\right)\frac{s_{\widehat t}^2}{m}
+\frac{M}{m}\sum_{h\in s}\widehat V_h.
$$

```r
set.seed(2207)
ids_2e <- unlist(lapply(clusters_s, function(h) {
  sample(marco_bci$id[marco_bci$conglomerado == h], l)
}))
s_2e <- marco_bci[match(ids_2e, marco_bci$id), ]

estimar_2e <- function(datos, variable, M, m, L, l, N_total) {
  por_h <- lapply(split(datos[[variable]], datos$conglomerado), function(z) {
    c(total_h = L * mean(z),
      var_h = L^2 * (1 - l / L) * var(z) / l)
  })
  por_h <- do.call(rbind, por_h)
  total <- M / m * sum(por_h[, "total_h"])
  var_total <- M^2 * (1 - m / M) * var(por_h[, "total_h"]) / m +
    M / m * sum(por_h[, "var_h"])
  c(total = total, var_total = var_total, media = total / N_total,
    EE_media = sqrt(var_total) / N_total)
}

e2_y <- estimar_2e(s_2e, "tallos", M, m, L, l, N_bci)
e2_y2 <- estimar_2e(transform(s_2e, tallos2 = tallos^2), "tallos2",
                    M, m, L, l, N_bci)
e2_p <- estimar_2e(s_2e, "riqueza_alta", M, m, L, l, N_bci)
S2_2e <- (e2_y2["total"] - (e2_y["total"]^2 - e2_y["var_total"]) / N_bci) /
  (N_bci - 1)
res_2e <- rbind(
  media_tallos = c(estimacion = unname(e2_y["media"]),
                   EE = unname(e2_y["EE_media"])),
  total_tallos = c(estimacion = unname(e2_y["total"]),
                   EE = unname(sqrt(e2_y["var_total"]))),
  proporcion_riqueza_alta = c(estimacion = unname(e2_p["media"]),
                              EE = unname(e2_p["EE_media"]))
)
res_2e <- cbind(res_2e, LI = res_2e[, "estimacion"] - t_cl * res_2e[, "EE"],
                LS = res_2e[, "estimacion"] + t_cl * res_2e[, "EE"])
res_2e["proporcion_riqueza_alta", c("LI", "LS")] <- pmin(1, pmax(0,
  res_2e["proporcion_riqueza_alta", c("LI", "LS")]))
round(res_2e, 3)
c(DE_tallos = sqrt(max(0, S2_2e)), hectareas_observadas = nrow(s_2e))
```

La segunda etapa añade el segundo componente de varianza: al muestrear tres de
cinco hectáreas por franja, el EE de la media sube a 9.58 (frente a 5.44
censando las cinco), porque además de la variación entre franjas hay que
estimar cada total de franja con una submuestra. El segundo componente
desaparece si se censan las cinco hectáreas. Con pocas unidades primarias, el
intervalo *t* sigue siendo aproximado aunque la varianza incorpore
correctamente ambas etapas.

## Información auxiliar: estimación de razón

Volvemos al MAS de diez árboles. El auxiliar
$x=\text{Girth}^2\text{Height}$ resume dimensiones relacionadas con volumen y se
conoce para las 31 unidades. La razón estimada y su linealización son

$$
\widehat R=\frac{\bar y_s}{\bar x_s},\qquad
e_i=y_i-\widehat R x_i,
$$

$$
\widehat{SE}(\widehat Y_R)\approx
N\sqrt{\left(1-\frac nN\right)\frac{s_e^2}{n}}.
$$

```r
X_aux <- sum(U$x_aux)
Rhat <- sum(s_mas$Volume) / sum(s_mas$x_aux)
Y_razon <- Rhat * X_aux
e_razon <- s_mas$Volume - Rhat * s_mas$x_aux
SE_razon <- N * sqrt((1 - n / N) * var(e_razon) / n)
IC_razon <- Y_razon + c(-1, 1) * t_mas * SE_razon

b_reg <- cov(s_mas$x_aux, s_mas$Volume) / var(s_mas$x_aux)
Y_reg <- N * (mean(s_mas$Volume) +
  b_reg * (mean(U$x_aux) - mean(s_mas$x_aux)))
e_reg <- s_mas$Volume - (coef(lm(Volume ~ x_aux, data = s_mas))[1] +
  b_reg * s_mas$x_aux)
SE_reg <- N * sqrt((1 - n / N) * var(e_reg) / n)

round(rbind(
  MAS = c(total = unname(est_mas["total"]), EE = unname(se_mas["total"]),
          LI = ic_mas["total", "LI"], LS = ic_mas["total", "LS"]),
  razon = c(total = Y_razon, EE = SE_razon, LI = IC_razon[1], LS = IC_razon[2]),
  regresion = c(total = Y_reg, EE = SE_reg,
                LI = Y_reg - t_mas * SE_reg, LS = Y_reg + t_mas * SE_reg)
), 2)
```

Con la única muestra MAS, la razón estima el total en 940.57 y la regresión en
938.98, ambos muy cerca de la verdad censal (935.3), mientras el MAS de esa
misma muestra da 1022.69, unas 87 unidades por encima. La ganancia está en la
incertidumbre: el EE del MAS es 156.50 y tanto la razón (20.60) como la
regresión (20.39) lo reducen a cerca de la séptima parte. Las pendientes casi
coinciden (0.0021 y 0.0022), señal de que el volumen es, en efecto, casi
proporcional al auxiliar (correlación 0.99). El estimador de regresión se
muestra como diagnóstico: permite intercepto, mientras la razón impone
proporcionalidad al origen. Ambos EE son aproximaciones por linealización.

### Gráfica 1: forma de la relación auxiliar

```r
#| label: fig-razon-relacion
#| fig-cap: "Volumen y auxiliar geométrico con ajustes de razón y regresión en la muestra MAS."
plot(U$x_aux, U$Volume, pch = 1, col = "grey55",
     xlab = expression(Diametro^2 %*% altura), ylab = "Volumen (pies cúbicos)")
points(s_mas$x_aux, s_mas$Volume, pch = 19, col = "#bc4749")
abline(a = 0, b = Rhat, col = "#386641", lwd = 2)
abline(lm(Volume ~ x_aux, data = s_mas), col = "#457b9d", lwd = 2, lty = 2)
legend("topleft", c("Razón", "Regresión"), col = c("#386641", "#457b9d"),
       lty = c(1, 2), lwd = 2, bty = "n")
```

La figura 2.17 muestra que el volumen depende del auxiliar con una relación
casi perfecta que parte del origen: los 31 puntos del censo se apiñan en torno
a una recta creciente (correlación 0.99). Sobre esa nube, la muestra (puntos
rojos) ajusta dos rectas casi coincidentes — la de razón (sólida, forzada al
origen) y la de regresión (discontinua, con intercepto). Como la dispersión
vertical es pequeña, los residuos son cortos y usar el auxiliar elimina gran
parte de la variabilidad que castigaba al MAS.

### Gráfica 2: residuos de proporcionalidad

```r
#| label: fig-razon-residuos
#| fig-cap: "Residuos del estimador de razón frente al auxiliar."
plot(s_mas$x_aux, e_razon, pch = 19, col = "#386641",
     xlab = expression(Diametro^2 %*% altura), ylab = "Residuo de razón")
abline(h = 0, lty = 2)
lines(lowess(s_mas$x_aux, e_razon), col = "#bc4749", lwd = 2)
```

La figura 2.18 examina los residuos de la razón frente al auxiliar: se reparten
alrededor de cero sin pendiente ni curvatura clara, y la suavizada (roja) apenas
se desvía de la línea horizontal. Eso confirma que la proporcionalidad describe
bien todo el rango y que no queda estructura que un estimador más flexible deba
rescatar. Una tendencia residual advertiría que una razón constante no describe
bien todo el rango; una correlación alta no basta, importa la forma y el
comportamiento de los residuos.

### Gráfica 3: precisión por repetición

```r
set.seed(2209)
rep_aux <- t(replicate(B, {
  z <- U[sample.int(N, n), ]
  r <- sum(z$Volume) / sum(z$x_aux)
  b <- cov(z$x_aux, z$Volume) / var(z$x_aux)
  c(MAS = N * mean(z$Volume),
    razon = r * X_aux,
    regresion = N * (mean(z$Volume) + b * (mean(U$x_aux) - mean(z$x_aux))))
}))
metricas_aux <- t(vapply(as.data.frame(rep_aux), metricas_repeticion,
                         numeric(3), verdad = verdad_arboles["total"]))
round(metricas_aux, 3)
```

```r
#| label: fig-razon-repeticion
#| fig-cap: "Distribuciones del total estimado mediante MAS, razón y regresión."
boxplot(as.data.frame(rep_aux), col = c("grey80", "#6a994e", "#457b9d"),
        ylab = "Total estimado de volumen")
abline(h = verdad_arboles["total"], lty = 2)
```

La figura 2.19 distribuye el total estimado en 3000 repeticiones. La caja del
MAS es muy ancha (DE 132.8): el total oscila lejos de la verdad (935.3, línea
discontinua). Las cajas de razón y regresión son estrechas (DE 20.4 y 22.8) y
centradas sobre la verdad. Con el auxiliar, la incertidumbre del total cae a
cerca de un sexto respecto del MAS.

### Gráfica 4: sesgo y RMSE

```r
#| label: fig-razon-metricas
#| fig-cap: "Sesgo absoluto y RMSE de los estimadores evaluados por repetición."
op <- par(mfrow = c(1, 2), mar = c(5, 4, 2, 1))
barplot(abs(metricas_aux[, "sesgo"]), names.arg = rownames(metricas_aux),
        col = "#bc4749", las = 2, ylab = "Sesgo absoluto")
barplot(metricas_aux[, "RMSE"], names.arg = rownames(metricas_aux),
        col = "#386641", las = 2, ylab = "RMSE")
par(op)
```

La figura 2.20 resume las dos métricas. El sesgo absoluto es pequeño en los
tres (regresión −0.10, razón −0.34, MAS −2.84), pero el RMSE lo domina la

razón asume un sesgo ligeramente mayor que la regresión a cambio del mejor
RMSE; la regresión, al estimar también el intercepto con solo diez puntos, paga
una pequeña penalización de varianza. La razón puede aceptar un sesgo pequeño a
cambio de una reducción importante de varianza. La comparación debe hacerse bajo
el diseño real y con un auxiliar que estuviera disponible antes de medir la
respuesta.

## Comparación y decisión

Las métricas siguientes reúnen los estimadores del total de volumen para la
población de 31 árboles. Conglomerados y dos etapas no se mezclan en esta tabla
porque estiman tallos en otra población y otra escala.

```r
comparacion_arboles <- rbind(
  MAS = c(estimacion = unname(est_mas["total"]),
          EE = unname(se_mas["total"]),
          sesgo_empirico = metricas_st["MAS", "sesgo"],
          DE_empirica = metricas_st["MAS", "DE_empirica"],
          RMSE = metricas_st["MAS", "RMSE"]),
  estratificado = c(estimacion = unname(est_st["total"]),
                    EE = unname(se_st["total"]),
                    sesgo_empirico = metricas_st["proporcional", "sesgo"],
                    DE_empirica = metricas_st["proporcional", "DE_empirica"],
                    RMSE = metricas_st["proporcional", "RMSE"]),
  sistematico = c(estimacion = unname(est_sys["total"]),
                  EE = unname(se_sys_diseno["total"]),
                  metricas_repeticion(todos_sys[, "total"],
                                      verdad_arboles["total"])),
  razon = c(estimacion = Y_razon, EE = SE_razon,
            sesgo_empirico = metricas_aux["razon", "sesgo"],
            DE_empirica = metricas_aux["razon", "DE_empirica"],
            RMSE = metricas_aux["razon", "RMSE"])
)
round(comparacion_arboles, 3)
```

La tabla ordena los cuatro estimadores por su RMSE de largo plazo, la métrica de
comparación imparcial. El MAS (RMSE 129.34) es la referencia: no aprovecha la
estructura del marco y paga la variabilidad completa de las unidades.
Estratificado (72.77) y sistemático (67.06) lo reducen casi a la mitad al
explotar, respectivamente, la homogeneidad de los estratos y la cobertura del
orden por diámetro; sistemático gana ligeramente porque el gradiente de volumen
con el diámetro es casi monótono. La razón (20.38) lo reduce a cerca de un sexto
al usar un auxiliar altamente correlacionado (0.99). Los sesgos empíricos son
pequeños en todos los diseños (−0.34 a 1.38): son insesgados o casi, y las
diferencias de RMSE provienen de la varianza, no del sesgo. No debe leerse solo
la columna de estimación puntual: el estratificado da 850.19 y el sistemático
1009.05, a lados opuestos de la verdad (935.3); una sola realización puede caer
a un lado u otro, y solo la repetición revela que sistemático es más estable.
Cada ganancia exige su condición operativa — estratos conocidos antes de
muestrear, orden y periodicidad estudiados, auxiliar disponible antes de medir —
y ninguno repara una selección no probabilística.

- **MAS** es la referencia cuando el marco es completo y no hay estructura útil.
- **Estratificación** es defendible si los estratos se conocen antes de muestrear
  y separan heterogeneidad o dominios importantes.
- **Sistemático** mejora cobertura, pero obliga a estudiar orden, periodicidad y
  estimación de varianza.
- **Conglomerados** reducen desplazamiento; su costo estadístico depende de la
  semejanza interna y del número de unidades primarias.
- **Dos etapas** equilibran cobertura dentro de grupos y costo, pero exigen
  conservar probabilidades y variación de ambas etapas.
- **Razón o regresión** aprovechan un marco auxiliar; no reparan una selección no
  probabilística ni una relación inestable.

Las comparaciones repiten mecanismos de selección sobre dos censos observados.
No demuestran que la clasificación de precisión se mantenga en otro bosque. En
un estudio nuevo deben conservarse marco, versiones de auxiliares, cortes de
estrato, identificadores de conglomerado, probabilidades, arranques, unidades,
costos y respuesta ante no observación.

La lista fija lo esencial para reproducir cualquier número del capítulo, y en

fija: los estimandos (media, total, proporción y DE), los umbrales de volumen
(30) y de riqueza (95), los cortes de estrato y la definición de los

ordenamiento no debe asumirse válido en otro bosque. En un estudio nuevo hay que

## Resultados breves de comprobación

**T1.** En los tres primeros casos el marco puede ser, respectivamente, la lista
o mapa de 500 cuadrantes, 80 colonias y 120 segmentos con sus puntos elegibles.
Las unidades primarias son cuadrante, colonia y segmento. En el cuarto caso la
unidad observada es un árbol accesible desde la carretera, pero no se conoce una
probabilidad positiva de inclusión para todos los árboles del bosque; un IC
convencional no corrige esa cobertura.

**T2.** Como $N\bar y_s=(N/n)\sum_iI_i y_i$ y $E(I_i)=n/N$, su esperanza es
$\sum_i y_i=Y$. Si $n=N$, no quedan muestras alternativas y el EE es cero. La DE
de las unidades no cambia al aumentar $n$; disminuye el EE del estimador.

**T3.** Las medias se ponderan por $W_h=N_h/N$. La ganancia de precisión requiere
estratos internamente homogéneos y diferentes entre sí. La representación de un
dominio puede ser un objetivo independiente de minimizar la varianza. Neyman
requiere DE previas, obtenidas de pilotos o estudios anteriores.

**T4.** Un gradiente suave suele favorecer la cobertura sistemática; un orden

## Actividad propuesta para el lector

Use el conjunto real `datasets::attitude` como población finita de 30 unidades y
estime la media, el total, la proporción que supera un umbral previamente fijado
y la DE de `rating`. Audite estructura, faltantes, duplicados, rangos y límites de
procedencia; defina unidad, marco y estimandos; y compare un MAS, una
estratificación previa por clases de `complaints`, y un sistemático sobre al menos
dos órdenes. Use un auxiliar justificado para un estimador de razón o regresión.
Para cada método documente probabilidades, pesos, estimador, EE e intervalo;
repita selecciones sobre el marco observado para examinar sesgo, DE empírica,
RMSE y cobertura; incluya al menos tres gráficas que expliquen selección,
heterogeneidad e incertidumbre; y evalúe sensibilidad al tamaño de muestra,
puntos de corte y orden. Finalmente, proponga cómo formar conglomerados y dos
etapas si las 30 unidades pertenecieran a grupos naturales, indicando qué
información adicional sería indispensable para ejecutar y analizar ese diseño.
